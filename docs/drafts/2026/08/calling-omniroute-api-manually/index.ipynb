```python
from openai import OpenAI
```

## Example of Calling OmniRoute Using the Python Library openai 

```python
client = OpenAI(
    base_url="http://localhost:20128/v1",
    api_key="YOUR_OMNIROUTE_API_KEY" # Generated in your local OmniRoute dashboard
)

response = client.chat.completions.create(
    model="auto/best-coding", # OmniRoute handles the routing behind the scenes
    messages=[
        {
            "role": "user",
            "content": "Write a Python function to reverse a string. Output the code ONLY."
        }
    ]
)
```
